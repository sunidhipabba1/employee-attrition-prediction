**Data-quality based feature removal**
1. **Constant features:** A feature with the same value for every observation has zero variance and cannot distinguish between observations. Therefore, EmployeeCount, StandardHours, and Over18 are removed.
2. **Identifier:** EmployeeNumber uniquely identifies employees but does not represent a meaningful employee characteristic. Including it could allow the model to learn arbitrary patterns associated with the identifier rather than meaningful relationships, so it is excluded.

```python
cols_to_drop = ['EmployeeCount', 'StandardHours', 'Over18', 'EmployeeNumber']
df = df.drop(columns=cols_to_drop)
df.shape
```

**Features requiring further investigation**
1. There seem to be three columns: DailyRate, HourlyRate and MonthlyRate, which need to be investigated further to check if they are giving the same information.
2. Understand whether the features MonthlyRate and MonthlyIncome are similar and correlated.

```python
df[['DailyRate', 'HourlyRate', 'MonthlyRate', 'MonthlyIncome']].corr()
```

DailyRate, HourlyRate, MonthlyRate, and MonthlyIncome show very low pairwise linear correlations, suggesting that they do not provide redundant linear information. Therefore, they will be retained for subsequent analysis.

# Data Preprocessing

```python
# Splitting the data into train and test folds
from sklearn.model_selection import train_test_split

X = df.drop(columns='Attrition')
y = df['Attrition'].map({'Yes':1,'No':0})

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print("Training shape:", X_train.shape)
print("Test shape:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTest target distribution:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

numerical_cols = X.select_dtypes(exclude='object').columns
categorical_cols = X.select_dtypes(include='object').columns

preprocessor = ColumnTransformer(
    transformers=[
        ('num',StandardScaler(),numerical_cols),
        ('cat',OneHotEncoder(handle_unknown='ignore', drop='first'),categorical_cols)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_pipeline = Pipeline([
    ('preprocessor',preprocessor),
    ('model',LogisticRegression(random_state=42))
])
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)

cv_results = cross_validate(logistic_pipeline,X_train,y_train,cv=cv,scoring=['accuracy','precision','recall','f1','roc_auc'])

cv_results_df = pd.DataFrame(cv_results)
cv_results_df
```

```python

metrics = ['test_accuracy','test_precision','test_recall','test_f1','test_roc_auc']

cv_summary = cv_results_df[metrics].agg(['mean','std']).T
cv_summary
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_pipeline = Pipeline([
    ('preprocessor',tree_preprocessor),
    ('model',DecisionTreeClassifier(random_state=42))
])
```

# Threshold Tuning the tuned LR

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_score,recall_score,f1_score

y_train_prob = cross_val_predict(lr_grid.best_estimator_,X_train,y_train,cv=cv,method='predict_proba')[:,1]
```

```python
thresholds = np.arange(0.3,0.65,0.05)
for threshold in thresholds:
  y_train_pred = (y_train_prob >= threshold).astype(int)
  print(f"Threshold:{threshold:.2f} | Precision:{precision_score(y_train,y_train_pred):.3f} |Recall:{recall_score(y_train,y_train_pred):.3f} | F1 score: {f1_score(y_train,y_train_pred):.3f}")
```

A threshold of 0.35 was selected instead of 0.40 because it provides higher recall (0.584 vs. 0.547) with only a small reduction in F1-score (0.594 vs. 0.599). Since identifying employees who may leave is an important objective, the higher recall was prioritized over the small improvement in F1-score at 0.40.

```python
best_model = lr_grid.best_estimator_
best_model.fit(X_train,y_train)

y_test_prob = best_model.predict_proba(X_test)[:,1]

threshold = 0.35
y_test_pred = (y_test_prob >= threshold).astype(int)
```

```python
from sklearn.metrics import accuracy_score,classification_report

print("Test Accuracy:",accuracy_score(y_test,y_test_pred))
print(classification_report(y_test,y_test_pred))
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(y_test, y_test_pred)

plt.figure(figsize=(6, 4))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['No Attrition', 'Attrition'],
    yticklabels=['No Attrition', 'Attrition']
)

plt.xlabel('Predicted Label')
plt.ylabel('Actual Label')
plt.title('Confusion Matrix — Tuned Logistic Regression')
plt.tight_layout()
plt.show()
```

**Confusion Matrix Interpretation**

The tuned Logistic Regression model correctly classified 228 employees who stayed and identified 23 employees who left. However, it missed 24 employees who actually left the organization, highlighting a limitation in detecting attrition cases. The model also generated 19 false positives. These results illustrate the trade-off between identifying employees at risk of leaving and incorrectly flagging employees who stay. Therefore, the predictions should support further investigation rather than be treated as definitive assessments of individual employees.

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

models = {
    'Tuned Logistic Regression': lr_grid.best_estimator_,
    'Decision Tree': tree_pipeline,
    'Random Forest': rf_pipeline,
    'XGBoost': xgb_pipeline
}

plt.figure(figsize=(8, 6))

for name, model in models.items():
    model.fit(X_train, y_train)

    probabilities = model.predict_proba(X_test)[:, 1]

    fpr, tpr, _ = roc_curve(y_test, probabilities)
    auc_score = roc_auc_score(y_test, probabilities)

    plt.plot(fpr, tpr, label=f'{name} (AUC = {auc_score:.3f})')

plt.plot([0, 1], [0, 1], 'k--', label='Random Classifier')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve Comparison')
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

# Conclusion

### 1. Model Selection

Tuned Logistic Regression achieved the highest F1-score among the evaluated models (approximately 0.572), while baseline Logistic Regression achieved the highest ROC-AUC (approximately 0.838). Tuned Logistic Regression was selected as the final candidate because it offered the best precision-recall balance according to F1-score. However, it did not perform best across every metric.

At the selected threshold, the model still missed some employees who left, so its predictions should be used as supporting information rather than definitive decisions.

A nested cross-validation approach could provide a less biased performance estimate, though the held-out test set serves as an independent evaluation in this analysis.

### 2. Key Findings and HR Recommendations

OverTime was the strongest predictor in the permutation importance analysis, followed by JobRole. Career progression, work experience, employee satisfaction, and BusinessTravel also contributed to predictions.

HR teams could investigate overtime patterns, role-specific retention challenges, career development opportunities, and employee satisfaction. These findings indicate associations, not proven causes of attrition.

### 3. Limitations

- Class imbalance makes attrition cases harder to identify.
- The model misses some employees who leave.
- Results from a single dataset may not generalize to other organizations.
- Feature importance and coefficients show associations, not causation.

### 4. Future improvements

Implement nested cross-validation to obtain a more robust estimate of model performance and further investigate ways to improve recall.
